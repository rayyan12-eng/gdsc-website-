Bell state generator

```python
import cirq

# Create two qubits
q0, q1 = cirq.LineQubit.range(2)

# Build circuit
circuit = cirq.Circuit(
    cirq.H(q0),          # Put q0 in superposition
    cirq.CNOT(q0, q1),   # Entangle q0 and q1
    cirq.measure(q0, q1, key='result')
)

# Simulate
sim = cirq.Simulator()
result = sim.run(circuit, repetitions=10)
print(circuit)
print(result)
```

Quantum Teleportation Protocol

```python
import cirq

# Qubits: sender, entangled partner, receiver
q0, q1, q2 = cirq.LineQubit.range(3)

# Circuit
circuit = cirq.Circuit()

# Prepare state to teleport
circuit.append(cirq.H(q0))
circuit.append(cirq.rx(0.5)(q0))

# Create entanglement between q1 and q2
circuit.append([cirq.H(q1), cirq.CNOT(q1, q2)])

# Bell measurement on q0 and q1
circuit.append([cirq.CNOT(q0, q1), cirq.H(q0)])
circuit.append([cirq.measure(q0, key='m0'), cirq.measure(q1, key='m1')])

# Conditional operations
circuit.append(cirq.CNOT(q1, q2).with_classical_controls('m1'))
circuit.append(cirq.CZ(q0, q2).with_classical_controls('m0'))

# Simulate
sim = cirq.Simulator()
result = sim.run(circuit, repetitions=1)
print(circuit)
print(result)
```

Quantum Fourier Transform (QFT) on 3 Qubits

```python
import cirq
import math

# Qubits
qubits = cirq.LineQubit.range(3)

# QFT function
def qft(qs):
    n = len(qs)
    circuit = cirq.Circuit()
    for i in range(n):
        circuit.append(cirq.H(qs[i]))
        for j in range(i+1, n):
            circuit.append(cirq.CZ(qs[j], qs[i])**(1/2**(j-i)))
    for i in range(n//2):
        circuit.append(cirq.SWAP(qs[i], qs[n-i-1]))
    return circuit

# Build and simulate
circuit = cirq.Circuit(qft(qubits), cirq.measure(*qubits, key='result'))
sim = cirq.Simulator()
result = sim.run(circuit, repetitions=5)
print(circuit)
print(result)
```
